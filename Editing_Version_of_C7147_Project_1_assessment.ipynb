## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
def validate_read(sequence, reference):
    """Check read and reference length. Check that reads contain valid characters and convert to upper-case."""
    errors = []
    if len(sequence) != len(reference):
      errors.append("wrong_length")

    sequence = sequence.upper()
    if not all(base in "ATGC" for base in sequence):
      errors.append("invalid_characters")

    return {"sequence": sequence, "valid": not errors, "errors": errors}
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
  """Return mutations position and reference base and read base"""
  mutations = []

  for i in range(len(sequence)):
    if sequence[i] != reference[i]:
      mutations.append({"position": i + 1, "reference_base": reference[i], "read_base": sequence[i]})

  return mutations
````

### 🔎 Expected outputs

- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
  """Validate the reads and find the mutations in invalid forming a list for each."""
  analysed_reads = []

  for read in reads:
    validation = validate_read(read["sequence"], reference)

    if validation["valid"]:
      status = "valid"
      mutations = find_mutations(validation["sequence"], reference)
    else:
      status = "invalid"
      mutations = []

    analysed_reads.append({"read_id": read["read_id"], "sample": read["sample"], "group": read["group"],
                            "sequence": validation["sequence"], "status": status, "errors": validation["errors"],
                            "mutations": mutations})

  return analysed_reads



````

````python
def summarise_samples(analysed_reads):
    """Sumarise samples as indiviudally from each group and return."""

    summaries = {}

    for item in analysed_reads:
        sample = item["sample"]

        if sample not in summaries:
            summaries[sample] = {"group": item["group"], "valid_reads": 0, "invalid_reads": 0,
                           "mismatch_reads": 0, "mismatch_percentage": None,
                           "mismatch_events": 0, "substitution_counts": {},
                           "position_counts": {}, "top_positions": []}

        data = summaries[sample] # shorten indiviudal summary -> data

        if item["status"] == "invalid": # Add valid/invalid reads to count
            data["invalid_reads"] += 1
        else:
            data["valid_reads"] += 1

            if len(item["mutations"]) > 0: # check for any mismatches. Empty = no mutations
              data["mismatch_reads"] += 1 # add into the mismatch_reads count
            data["mismatch_events"] += len(item["mutations"]) # count indiviudal mismatch events. 3 mismatches = 3 individuals
            data["mismatch_percentage"] = (data["mismatch_reads"] / data["valid_reads"]) * 100 # mismatch / valid * 100 give a percentage of mismatches

            for mutation in item["mutations"]: # for every mismatch

                change = mutation["reference_base"] + ">" + mutation["read_base"] # join ref and read bases create substitution label showing a mismatch
                data["substitution_counts"][change] = data["substitution_counts"].get(change, 0) + 1 # Count for how often substitutions occur in indiviudal sample e.g. how many A>G occur (ref has A, read has G)

                position = mutation["position"] # find the current mismatch position from find_mutation dictionary
                data["position_counts"][position] = data["position_counts"].get(position, 0) + 1 # position count for current mismatch


    for sample in summaries:
        data = summaries[sample]

        ordered_positions = sorted(data["position_counts"]) # list mismatch positions smallest first
        ordered_positions.sort(key = data["position_counts"].get, reverse = True) # reorder by mismatch count with highest first
        data["top_positions"] = ordered_positions[0:3] # slice first 3 items of ordered_positions (the top 3)

    print_sample_tables(summaries)
    return summaries



def print_sample_tables(summaries):
    """  """
    samples_by_group = {} # empty dictionary for groups
    for sample in summaries:
        data = summaries[sample]
        group = data["group"]

        if group not in samples_by_group: # if group is the first encounter
            samples_by_group[group] = [] # list to hold names of samples belonging to specific groups
        samples_by_group[group].append(sample) # add the current sample to groups[group]

    for group in samples_by_group:
        print(f"Group: {group}") # print out the value of group e.g. treatment
        print("Sample | Valid reads | Invalid reads | Mismatch % | Mismatch events | Top positions") # summary table headings

        for sample in samples_by_group[group]:
            data = summaries[sample] # shorten indiviudal summary
            if data["mismatch_percentage"] is None: # if no percentage print n/a
              percentage_text = "N/A"
            else:
                percentage_text = f'{data["mismatch_percentage"]:.2f}%' # print mismatch percentage to 2 d.p.
            print(sample, data["valid_reads"], data["invalid_reads"], percentage_text, data["mismatch_events"], data["top_positions"], sep="|") # print one row for this sample seperate by |

        substitution_types = []
        for sample in samples_by_group[group]: # pick a sample
            data = summaries[sample]
            for change in data["substitution_counts"]: # pick each subsitution label from samples dictionary
                if change not in substitution_types:
                    substitution_types.append(change)
        print("\nSubstitutions (count and events per 100 valid reads)")
        print(" | ".join(["Substitution"] + samples_by_group[group]))

        if len(substitution_types) == 0:
            print("No substitutions observed.")

        for change in sorted(substitution_types):
            row = [change]
            for sample in samples_by_group[group]:
                data = summaries[sample]
                count = data["substitution_counts"].get(change, 0)
                if data["valid_reads"] == 0:
                    row.append(f"{count} (N/A)")
                else:
                    percentage = count / data["valid_reads"] * 100
                    row.append(f"{count} ({percentage:.2f}%)")
            print(" | ".join(row))

        print("N/A means the sample has no valid reads.")







````

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:


Each list must contain **one percentage per biological sample**.

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````
